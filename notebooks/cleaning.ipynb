```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("../data/credit_risk_dataset_raw.csv")
df.head(20)
```

```python
df = df.convert_dtypes()
```

```python
for column in df.columns:
    if df[column].dtype == "string":
        df[column] = df[column].astype("object")
```

```python
df[df["person_age"] > 100]
```

```python
df["person_age"].describe()
```

```python
df = df.drop(df[df["person_age"] > 100].index)
```

```python
df[df["loan_int_rate"].isnull()]
```

```python
df[df["person_emp_length"].isnull()]
```

```python
imputer = SimpleImputer(strategy="median")
imputer.fit(df[["loan_int_rate", "person_emp_length"]])
df[["loan_int_rate","person_emp_length"]] = imputer.transform(df[["loan_int_rate","person_emp_length"]])
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df["person_emp_length"] = df["person_emp_length"].astype("Int64")
df["loan_int_rate"] = df["loan_int_rate"].astype("Float64")
```

```python
df.info()
```

```python
df.describe().transpose()
```

```python
df[df["person_emp_length"] > 30]
```

```python
df.drop(df[df["person_emp_length"] > 100].index, inplace=True)
```

```python
df["loan_status"] = df["loan_status"].astype("object")
```
